```python
import numpy as np

class SimpleNN:
    def __init__(self, n_in, n_hidden, n_out, lr=0.1):
        self.lr = lr
        self.W1 = np.random.randn(n_in, n_hidden) * 0.1
        self.b1 = np.zeros(n_hidden)
        self.W2 = np.random.randn(n_hidden, n_out) * 0.1
        self.b2 = np.zeros(n_out)

    def _sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def _sigmoid_grad(self, y):
        return y * (1 - y)

    def forward(self, X):
        self.a1 = self._sigmoid(X.dot(self.W1) + self.b1)
        self.a2 = self._sigmoid(self.a1.dot(self.W2) + self.b2)
        return self.a2

    def train(self, X, y, epochs=10000, print_every=2000):
        history = {'epochs': [], 'loss': []}
        
        for ep in range(1, epochs+1):
            # Прямой проход
            out = self.forward(X)
            
            # Вычисление ошибки
            loss = np.mean((out - y)**2)
            
            # Обратное распространение ошибки
            e2 = (out - y) * self._sigmoid_grad(self.a2)
            e1 = e2.dot(self.W2.T) * self._sigmoid_grad(self.a1)
            
            # Обновление весов
            self.W2 -= self.lr * self.a1.T.dot(e2)
            self.b2 -= self.lr * e2.sum(axis=0)
            self.W1 -= self.lr * X.T.dot(e1)
            self.b1 -= self.lr * e1.sum(axis=0)
            
            # Сохраняем историю обучения
            if ep % print_every == 0 or ep == epochs:
                history['epochs'].append(ep)
                history['loss'].append(loss)
                print(f"Эпоха {ep}/{epochs}, MSE Loss={loss:.4f}")
        
        # Финальные предсказания
        final_outputs = self.forward(X)
        print("\nРезультаты XOR:")
        print(f"{'Вход':<12} | {'Вероятность':<12} | {'Предсказание':<12} | {'Целевое':<12}")
        print("-" * 55)
        
        for xi, yi, oi in zip(X, y, final_outputs):
            prob = float(oi)
            pred_bin = 1 if prob > 0.5 else 0
            correct = "✓" if pred_bin == int(yi) else "✗"
            print(f"{str(xi.tolist()):<12} | {prob:.6f}    | {pred_bin:<12} | {int(yi):<8} {correct}")
        
        return final_outputs, history

# Демонстрация на задаче XOR
if __name__ == '__main__':
    # Данные XOR
    X = np.array([[0,0], [0,1], [1,0], [1,1]])
    y = np.array([[0], [1], [1], [0]])
    
    # Инициализация и обучение модели
    nn = SimpleNN(n_in=2, n_hidden=4, n_out=1, lr=0.5)
    outputs, history = nn.train(X, y, epochs=10000, print_every=2000)
    
```
